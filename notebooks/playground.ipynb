```python
import glob
import pandas as pd
csv_file_path= "/scratch/akash/metadata"
csv_files= glob.glob(csv_file_path+"/*.csv")

print(len(csv_files))
total_count= 0
high_sim_count= 0
for file in csv_files:
    df= pd.read_csv(file, engine="python", on_bad_lines="skip")
    # Convert the similarity score column to numeric, coercing non-numeric to NaN
    df["clip_b32_similarity_score"] = pd.to_numeric(df["clip_b32_similarity_score"], errors="coerce")
    df_high_sim= df[df["clip_b32_similarity_score"]>0.28]
    total_count+= len(df)
    high_sim_count+= len(df_high_sim)

print(total_count)
print(high_sim_count)
```
